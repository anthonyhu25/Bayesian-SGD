#1 Problem Setup


This [paper](https://arxiv.org/pdf/2202.07581) is taking a Bayesian Stochastic Gradient Descent (SGD) approach towards stochastic optimization of a batch of streaming data every time interval, in the decision-indepedent and decision-independent case. In stochastic simulation optimization, there has been much work done on decision-independent (exogenous) data, where incoming data is independent to decision-making by other parties. In reality, these assumptions are unrealistic and infeasible for solving decision-dependent (endogenous) data optimization problems.

This paper addresses cases where data from the distribution (in both decision-independent and decision-dependent) cases arrive in batches over time, and the underlying distribution is unknown. In both cases, Bayesian estimation provides challenges with the updates on the distributions due to the consistency of the Bayesian posterior distribution, not to mention the correlation and non-stationarity of the decision-dependent data.

For the simulated example, I will start with the decision-independent example, and recreate the figures in the decision-independent cases in the paper.

## 1.1 Decision-Independent Case

We are attempting to maximize the objective function

$\large min_{x\in \chi} H(x) := 𝔼_{f(.;\theta^{c})} [h(x,\xi)]$

Where $x \in \chi ⊂ ℝ ^ {d}$ is the decision vector, $ξ \in \Xi \subset ℝ ^{m}$ is a random vector, $h: ℝ^{d} * ℝ^{m} ↦ ℝ$ is a deterministic function. Note that $\xi \sim f(.|x, θ^{c})$ and we denote the prior on parameter $\theta_{c}$ on time step $t$ as $\pi_{t}(\theta)$

Main idea of the paper is assuming the batch of data at time $t = 1,...,T$ with batch size  $D$, denoted as ${y_{1,t},...,y_{D,t}} \stackrel{iid}{\sim} f(.|\theta^{c})$. We start off with initial decision $x_{0}$ and initial prior on $\theta$, denoted as $\pi_{0}(\theta)$.

For the stochastic gradient descent update, we update the batch of decisions for time $t$, denoted as ${x_{1,t},...,x_{D,t}}$ and with stepsize $\alpha_{t,j}$ for $j = 1,...,D$.

The SGD gradient estimator will be the gradient with respect to the objective function $h(x,ξ)$, denoted as $∇_{x}h(x,\xi)$. So, with each batch of decisions ${x_{1,t},...,x_{D,t}}$ for each timestamp $t$, we will update the final decision for the next batch of streaming data at timestamp $t+1$. The update rule for the batch follows:

1. Set decision for batch at time $t$ -- $x_{t}$ as $x_{t,1}$
2. While $1 \le j \le D$, set $x_{t,j+1} = x_{t,j} -\alpha_{t,j} ∇_{x}h(x_{t,j}, ξ_{t,j})$ where $\xi_{t,j}$ is generated from $f(.|\theta^{c}_{t,j})$, such that $\theta^{c}_{t,j}$ is generated from the posterior at time step $t$, $\pi_{t}(\theta)$

Note that the posterior $\pi_{t}(\theta)$ is calculated with respect to the data $\textbf{y}_{t} = {{y_{1,t},...,y_{D,t}}}$, and posterior for previous timestep $\pi_{t-1}(\theta)$.

$\large \pi_{t}(\theta) = \frac{\pi_{t-1}(\theta)f(\textbf{y}_{t}|\theta_{c})}{\int \pi_{t-1}(\theta)f(\textbf{y}_{t}|\theta_{c}) dθ}$

where $\large f(\textbf{y}_{t}|\theta_{c}) = Π^{D}_{j=1}f(y_{t,j}|\theta)$
